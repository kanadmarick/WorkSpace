##### Pydantic basics: Creating and Using Models
Pydantic is a data validation and settings management library for Python, based on Python type annotations. It allows you to define data models with type hints, and it automatically validates the data against those types. 


```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name: str
    age: int
    city: str

person=Person(name="Kanad", age=30, city="New York")
type(person)
print(person)    
```

```python
from typing import Optional
```

```python
class Employee(BaseModel):
    name: str
    age: int
    city: str
    department: str
    salary: Optional[float] = None  # Optional field with a default value of None   
    is_active: Optional[bool] = True  # Optional field with a default value of True
    
```

```python

employee1= Employee(name="Alice", age=28, city="Los Angeles", department="Engineering", salary=75000.0)
print(employee1)
```

###Defination
1. Optional[type]: This is a type hint that indicates that a variable can either be of the specified type or None. It is used to define optional fields in Pydantic models. For example, if you have a field that can either be an integer or None, you would use Optional[int].
2. Default Value: In Pydantic models, you can provide default values for fields. If a field has a default value, it means that if the field is not provided when creating an instance of the model, it will automatically take on that default value. For example, if you have a field with a default value of 10, and you create an instance of the model without specifying that field, it will be set to 10 by default.
3. Pydantic Validates types even for optional fields: Pydantic performs type validation for all fields, including optional ones. If you specify a type for an optional field, Pydantic will check that the value provided (if any) matches the specified type. If the value does not match the expected type, Pydantic will raise a validation error. This ensures that even optional fields adhere to the defined data types when they are provided.
4.Required Fields: In Pydantic models, fields without default values are considered required. When creating an instance of the model, you must provide values for all required fields. If you omit a required field, Pydantic will raise a validation error indicating that the field is missing. This helps ensure that essential data is always provided when working with the model.


```python
from pydantic import BaseModel
from typing import List

class Classroom(BaseModel):
    name: str
    students: list[str]
    capacity: int
    is_active: Optional[bool] = True  # Optional field with a default value of True
```

```python
# Create a classroom
classroom = Classroom(name="Math 101", students=("Alice", "Bob", "Charlie"), capacity=30)
print(classroom)    
```

## Model with Nested Models

Create complex data structures by nesting Pydantic models within each other. This allows you to represent hierarchical data and validate it effectively.

```python
from pydantic import BaseModel

class Address(BaseModel):
    street: str
    city: str
    state: str
    zip_code: str

class Customer(BaseModel):
    name: str
    email: str
    address: Address  # Nested model    
    
# Creates a customer with a nested address
customer = Customer(
    name="John Doe",
    email="",
    address=Address(
        street="123 Main St",
        city="Anytown",
        state="CA",
        zip_code="12345"
    )
)   
print(customer)  
```

### Pydantic Fields: Customiziation and Constraints

The field functionin Pydantic enhances the capabilities of Pydantic models by allowing you to customize field behavior and apply constraints. It provides a way to define additional metadata, validation rules, and default values for model fields. By using the field function, you can enforce specific requirements on your data, such as minimum or maximum values, regex patterns, and more. This helps ensure that the data adheres to the desired structure and constraints, improving the overall reliability and integrity of your models. 

```python

from pydantic import BaseModel, Field
class Item(BaseModel):
    name: str = Field(..., description="The name of the item", min_length=1)
    price: float = Field(..., gt=0, description="The price of the item, must be greater than 0")
    quantity: int = Field(..., ge=0, description="The quantity of the item in stock, must be non-negative") 
    
item = Item(name="Laptop", price=999.99, quantity=10)
print(item)
    
```

```python
from pydantic import BaseModel, Field

class User(BaseModel):
    username: str =Field(..., description="The username of the user", min_length=3, max_length=20)
    age: int = Field(..., ge=0, description="The age of the user, must be non-negative")
    email: str = Field(default=lambda: "user@example.com", description="The email of the user")
    
#Examples

user1 = User(username="john_doe", age=25)
print(user1)

user2 = User(username="jane_smith", age=30, email="jane_smith@example.com")
print(user2)
```
